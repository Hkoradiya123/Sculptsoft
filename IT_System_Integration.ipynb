# IT System Integration using APIs

This notebook demonstrates how to integrate with IT systems using APIs for automation and data retrieval.

## 1. Set Up API Authentication

Configure API authentication using environment variables, API keys, or OAuth tokens to securely access IT systems.

```python
import os
import requests
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# API Authentication Configuration
class APIAuth:
    def __init__(self):
        # Load credentials from environment variables
        self.api_key = os.getenv('IT_API_KEY')
        self.api_secret = os.getenv('IT_API_SECRET')
        self.base_url = os.getenv('IT_BASE_URL', 'https://api.itsm.example.com')
        self.oauth_token = os.getenv('IT_OAUTH_TOKEN')
        
    def get_headers(self):
        """Generate authentication headers for API requests"""
        headers = {
            'Content-Type': 'application/json',
            'Accept': 'application/json'
        }
        
        # API Key Authentication
        if self.api_key:
            headers['Authorization'] = f'ApiKey {self.api_key}'
        
        # OAuth Token Authentication
        elif self.oauth_token:
            headers['Authorization'] = f'Bearer {self.oauth_token}'
            
        return headers
    
    def test_connection(self):
        """Test API connectivity"""
        try:
            response = requests.get(
                f'{self.base_url}/health',
                headers=self.get_headers(),
                timeout=10
            )
            return response.status_code == 200
        except Exception as e:
            print(f"Connection failed: {e}")
            return False

# Initialize authentication
auth = APIAuth()
print("API Authentication configured successfully")
print(f"Base URL: {auth.base_url}")
print(f"API Key configured: {'Yes' if auth.api_key else 'No'}")
print(f"OAuth Token configured: {'Yes' if auth.oauth_token else 'No'}")
```
